```python
def get_product(x: int, y: int) -> int:
    return x*y
```

```python
get_product(2, 3)
```

```python
get_product(2, "4")
```

To handle errors like in the above cell, call ensure()

```python
from ensure import ensure_annotations
```

```python
@ensure_annotations
def get_product(x: int, y: int) -> int:
    return x*y
```

```python
get_product(2, "3") # This will raise an error because of type checking
```

Output:  
EnsureError: Argument y of type <class 'str'> to <function get_product at 0x00000165336F07C0> does not match annotation type <class 'int'>
